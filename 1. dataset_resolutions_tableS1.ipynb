```python
import numpy as np
import pandas as pd

R = 6371.0088  # Earth radius in km

def latlon_cell_area(lat, resolution):
    """
    Approximate spherical area of a regular lat-lon grid cell.

    lat        : cell-center latitude in degrees
    resolution : grid spacing in degrees
    """

    dlat = np.deg2rad(resolution)
    dlon = np.deg2rad(resolution)

    lat1 = np.deg2rad(lat - resolution / 2)
    lat2 = np.deg2rad(lat + resolution / 2)

    area = R**2 * dlon * (np.sin(lat2) - np.sin(lat1))

    return area


latitudes = [0, 10, 20, 30, 40, 50, 60, 70]

datasets = {
    "ERA5": {
        "grid_spacing": "0.25°",
        "resolution_deg": 0.25,
        "reported_resolution": "~31 km"
    },
    "ERA5-Land": {
        "grid_spacing": "0.10°",
        "resolution_deg": 0.10,
        "reported_resolution": "~9 km"
    },
    "MODIS snow": {
        "grid_spacing": "0.05°",
        "resolution_deg": 0.05,
        "reported_resolution": "~5 km"
    },
    "GLAD cropland": {
        "grid_spacing": "0.025°",
        "resolution_deg": 0.025,
        "reported_resolution": "~3 km"
    },
    "GRACE/GRACE-FO": {
        "grid_spacing": "0.5° distributed grid",
        "resolution_deg": 0.5,
        "reported_resolution": "~3° effective spatial resolution; distributed on a 0.5° grid"
    }
}

rows = []

for dataset, info in datasets.items():

    row = {
        "Dataset": dataset,
        "Native grid spacing": info["grid_spacing"],
        "Reported/effective spatial resolution": info["reported_resolution"]
    }

    for lat in latitudes:
        area = latlon_cell_area(lat, info["resolution_deg"])
        row[f"Cell area at {lat}°N (km²)"] = round(area, 1)

    rows.append(row)


df_resolution = pd.DataFrame(rows)

df_resolution



# table = {}

# for dataset, res in datasets.items():
#     table[dataset] = [
#         latlon_cell_area(lat, res)
#         for lat in latitudes
#     ]

# df = pd.DataFrame(
#     table,
#     index=[f"{lat}°N" for lat in latitudes]
# ).T

# df = df.round(1)

# df
```
